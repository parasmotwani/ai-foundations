### A lambda function is a small anonymous function.

```python
def lowercase_text(text):
    return text.lower()

# Lambda equivalent is

lowercase_text2 = lambda text: text.lower() # lambda parameters: expression

# Both can be called same way

lowercase_text("HeLLo")
lowercase_text2("HeLLo")
```

`lambda x: x * 2`

means:
Take x and return x * 2.

Unlike `def`, you don't explicitly write `return`. The expression's result is automatically returned.

```python
square_number = lambda x: x * x
square_number(12)
square_number(5)

add_tax = lambda price: price + ((price * 18) / 100)
add_tax(100)
```

#### `map()` + `lambda`

```python
requests = [
    {"model": "gpt", "tokens": 1200},
    {"model": "gpt", "tokens": 500},
    {"model": "gpt", "tokens": 3200},
    {"model": "gpt", "tokens": 800},
]


# map(     
#     WHAT TO DO,
#     WHAT TO DO IT TO
# )
#
# i.e.
#
# map(function, iterable)

tokens = map(                           # map() is a built-in funtion
    lambda request: request["tokens"],
    requests
)
tokens = list(tokens)
print(tokens)
```

```python
requests = [
    {"model": "gpt", "tokens": 1200},
    {"model": "gpt", "tokens": 500},
    {"model": "gpt", "tokens": 3200},
    {"model": "gpt", "tokens": 800},
]

costs = map(
    lambda request: ((request["tokens"] / 1000) * 10), # $10 per 1,000 tokens
    requests
)
print(costs) # map object

costs = list(costs)
print(costs)  # actual list
```

```python
def calculate_cost(request):
    return request["tokens"]/1000 * 10

costs = map(
    calculate_cost,
    requests
)
print(list(costs))
```

#### `filter()` + `lambda`

```python
# filter(
#     WHAT CONDITION TO CHECK,
#     WHAT TO CHECK IT AGAINST
# )
# JUST LIKE map() BUT USED FOR FILTERING

requests = [
    {"model": "gpt", "tokens": 1200},
    {"model": "gpt", "tokens": 500},
    {"model": "gpt", "tokens": 3200},
    {"model": "gpt", "tokens": 800},
]

expensive_requests = filter(
    lambda request: request['tokens'] > 1000,
    requests
)

print(list(expensive_requests))

cheap_requests = filter(
    lambda request: request['tokens'] < 1000,
    requests
)

print(list(cheap_requests))
```

`map()`
→ every item goes in
→ every item produces a result
→ number of results = number of input items

`filter()`
→ every item is tested
→ only True items remain
→ number of results ≤ number of input items

#### `sorted()` + `lambda`

```python
numbers = [50, 10, 30, 20]

sorted(numbers)
```

```python
requests = [
    {"model": "gpt", "tokens": 1200},
    {"model": "gpt", "tokens": 500},
    {"model": "gpt", "tokens": 3200},
    {"model": "gpt", "tokens": 800},
]

sorted_requests = sorted(requests,
       key = lambda request: request['tokens'])
print(sorted_requests)

sorted_requests_desc = sorted(requests,
       key = lambda request: request['tokens'],
       reverse = True)
print(sorted_requests_desc)
```

A lambda is simply a function expression, and because functions are first-class objects, you can pass that function wherever another function expects one.

Use a `lambda` for a small, simple function that is usually used temporarily and doesn't need a meaningful name. Use `def` when the function is reusable, complex, or deserves a descriptive name.
