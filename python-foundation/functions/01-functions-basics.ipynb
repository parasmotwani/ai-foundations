```python
def subtotal(price: int, quantity: int) -> int:  # Variables defined are parameters
    value = price * quantity
    return value

price = 1000 # Values being passed to a function is an argument
quantity = 10

subtotal(price, quantity)
```

```python
def create_order(product:str = "Laptop", unit_price:int = 75000, quantity:int = 1) -> list: # Variables defined are parameters
    order = [product, unit_price, quantity]
    return order

create_order()

create_order(
    product = "Mobile", # Values being passed to a function is an argument
    unit_price = 60000,
    quantity = 1
)

create_order("perfume", 1200, 3)

create_order("PC", unit_price=1888830, quantity=2)
```

```python
def estimate_request_cost(input_tokens:int, output_tokens:int, input_per1k_cost:int, output_per1k_cost:int) -> float:
    input_cost = (input_tokens * input_per1k_cost) / 1000
    output_cost = (output_tokens * output_per1k_cost) / 1000
    total_cost = input_cost + output_cost
    return total_cost

estimate_request_cost(100000, 2000, 10, 30)
```

```python
def configure_model(model: str, **options) -> dict: # **kwargs allows a function to accept a variable number of keyword arguments.
    config = {
        "model": model,
        **options
 }
    return config
    
configure_model(
    "gpt-model",
    temperature=0.7,
    max_tokens=500,
    streaming=True
)

configure_model(
    "embedding-model",
    dimensions=1536,
    normalize=True
)
```

```python
# *args lets a function accept a variable number of positional arguments.

# Important distinction

# The * in the function definition means:

# Collect all additional positional arguments into one tuple.

# It does not mean that you're passing a tuple manually.

def log_event(message, *details): # *args collects positional arguments into tuples
    log = {
        "message": message,
        "details": details
    }
    return log

log_event(
    "LLM request completed",
    "gpt-model",
    1500,
    0.42
)

log_event(
    "Tool execution failed",
    "search_web",
    504,
    "timeout"
)

log_event("Agent started")
```

*args   → positional arguments → tuple

**kwargs → keyword arguments → dictionary

but args and kwargs are just conventional names. The * and ** are what matter.

```python
def create_llm_request(model: str, *messages, **options) -> dict:
    request = {
        "model": model,
        "messages": messages,
        "options": options
    }
    return request

create_llm_request(
    "gpt-model",
    "System: You are helpful.",
    "User: Explain Python.",
    temperature=0.7,
    max_tokens=500,
    streaming=True
)
```
