You're building a simple AI agent pipeline.
Each agent step is represented by a function:

`def clean_input(text):
    return text.strip()`

`def lowercase(text):`
    `return text.lower()`

`def add_label(text):`
    `return f"USER INPUT: {text}"`

Your job is to build a reusable build_pipeline() function that:
- accepts any number of processing functions using *args
- returns a new function
- the returned function applies each processor in order
- passes the output of one processor into the next
So this:

`pipeline = build_pipeline(
    clean_input,
    lowercase,
    add_label
)`

should create a function that can then be used like:

`pipeline("   HELLO WORLD   ")`

Expected result:
`USER INPUT: hello world`

```python
def clean_input(text):
    return text.strip()

def lowercase(text):
    return text.lower()

def add_label(text):
    return f"USER INPUT: {text}"

def build_pipeline(*processors):
    def pipeline(text):
        for process in processors:
            text = process(text)
    
        return text
    return pipeline

pipeline = build_pipeline(
    clean_input,
    lowercase,
    add_label
)
pipeline("   HELLO WORLD   ")
```
