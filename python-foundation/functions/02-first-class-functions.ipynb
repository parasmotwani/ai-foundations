### A function is an object

```python
def clean_text(text: str) -> str:
    return text.strip().lower()

def normalize_text(text: str) -> str:
    return text.strip().upper()

def remove_trailing_spaces(text: str) -> str:
    return text.strip()

processor = clean_text # processor doesn't contain the result of 'clean_text' it contains a reference to the function itself
processor("   The Agent Needs Clean Input   ")

processor = normalize_text
processor("   The Agent Needs Clean Input   ")
```

### The key distinction

`processor = clean_text`

stores/references the function.



`processor = clean_text()`

calls the function and stores its return value.

```python
def process_text(text, processor):
    return processor(text)

process_text(                                   # Calls the function passed as the processor argument
    "   Agent Tools Need Consistent Input   ",
    clean_text
)

process_text(
    "   Agent Tools Need Consistent Input   ",
    normalize_text
)

process_text(
    "   Agent Tools Need Consistent Input   ",
    remove_trailing_spaces
)
```

```python
def get_processor(mode):
    if mode == "clean":
        return clean_text
    elif mode == "normalize":
        return normalize_text
    elif mode == "remove_spaces":
        return remove_trailing_spaces
    else:
        return "Provide a valid mode"


# get_processor() selects and returns a function based on the mode.
processor = get_processor("clean")

# The returned function can then be called through the processor variable.
processor("   Hello WORLD   ")


processor = get_processor("normalize")
processor("   Hello WORLD   ")


processor = get_processor("remove_spaces")
processor("   Hello WORLD   ")
```

```python
def create_text_processor(mode):
    if mode == "clean":
        return clean_text
    elif mode == "normalize":
        return normalize_text
    elif mode == "remove_spaces":
        return remove_trailing_spaces


# The function returns another function based on the selected mode.
processor = create_text_processor("clean")

# processor now references the function returned by create_text_processor().
result = processor("   Agent Tools Need Consistent Input   ")

print(result)
```

```python
def lowercase_text(text):
    return text.lower()


def remove_spaces(text):
    return text.strip()


def process_pipeline(text, *processors):
    # Apply each processor sequentially, passing the previous result to the next processor.
    for process in processors:
        text = process(text)

    # Return the final result after all processors have been applied.
    return text


process_pipeline(
    " Hello WORLD    ",
    lowercase_text,
    remove_spaces
)
```
