A callback is simply a function that you pass to another function so that the receiving function can call it at some point.

There isn't new syntax here. It's a pattern built from first-class functions.

### Callback

```python
def handle_request(request, on_complete):
    result = request()
    return on_complete(result) # on_complete(result) is a callback

    # The function receiving the callback decides when to call it.

def send_request():
    return "LLM response received"

def log_result(result):
    return f"LOG: {result}"

handle_request(send_request, log_result)
```

```python
def execute_tool(tool, on_success):
    result = tool()
    return on_success(result)

def search_database():
    return "Found 3 matching documents"

def format_result(result):
    return f"TOOL RESULT: {result}"

execute_tool(search_database, format_result)
```

### Function Composition

```python
# Take the output of one function and use it as the input of another function.

def clean_text(text: str) -> str:
    return text.strip()

def normalize_text(text: str) -> str:
    return text.strip().lower()

def compose(text, first, second):
    result = first(text)
    return second(result)

process = compose(
    "       Hi There   ",
    clean_text,
    normalize_text
)

print(process)
```

```python
def remove_extra_spaces(text):
    return " ".join(text.split())


def add_prefix(text):
    return f"LLM RESPONSE: {text}"

def compose(first, second):
    def composed(text):
        result = first(text)
        result = second(result)
        return result
    return composed

    

process = compose(
    remove_extra_spaces,
    add_prefix
)

process("Hi")
```
