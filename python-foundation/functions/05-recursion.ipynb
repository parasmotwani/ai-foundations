Recursion is when a function calls itself.

```python
def countdown(number):
    if number == 0:
        return
    
    print(number)
    countdown(number - 1)

countdown(2)
```

```python
def countdown(number):
    if number == 0:
        return []

    return [number] + countdown(number - 1)

countdown(10)

# countdown(3)
# → [3] + countdown(2)
# → [3] + [2] + countdown(1)
# → [3] + [2] + [1] + countdown(0)
# → [3] + [2] + [1] + []
# → [3, 2, 1]
```

```python
tasks = {
    "name": "Build RAG system",
    "subtasks": [
        {
            "name": "Prepare documents",
            "subtasks": [
                {
                    "name": "Parse PDFs",
                    "subtasks": []
                },
                {
                    "name": "Create chunks",
                    "subtasks": []
                }
            ]
        },
        {
            "name": "Build retrieval",
            "subtasks": [
                {
                    "name": "Generate embeddings",
                    "subtasks": []
                }
            ]
        }
    ]
}

def get_task_names(task):
    if not task["subtasks"]:
        return [task["name"]]
    
    names = [task['name']]

    for subtask in task["subtasks"]:
        names += get_task_names(subtask)

    return names

result = get_task_names(tasks)
print(result)

# def function(item):
#     if item_has_no_children:
#         return result

#     result = current_item

#     for child in children:
#         result += function(child)

#     return result
```
