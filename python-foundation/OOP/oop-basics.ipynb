```python
class LLMModel:
    def __init__(self, name, temperature): # __init__ is used to initialize an object when it is created.
        self.name = name
        self.temperature = temperature

model = LLMModel("GPT", 0.7)

print(model.name)
print(model.temperature) 

# name and temperature are called attributes

# When you do:
# model = LLMModel("GPT", 0.7)


# Python automatically runs the initialization logic.

# LLMModel("GPT", 0.7)
#         ↓
# __init__(...)
#         ↓
# self.name = "GPT"
# self.temperature = 0.7
```

## Methods — behavior belonging to an object

We can put functions inside a class.
Inside a class, these are called methods.

```python
class LLMModel:
    def __init__(self, name, temperature):
        self.name = name
        self.temperature = temperature

    def describe(self):
        return f"{self.name} runs at temperature {self.temperature}"
    
model = LLMModel("GPT", 0.7)
model.describe()

# Notice:
# model.describe()


# rather than:
# describe(model)


# The method belongs to the object.
```

```python
# The mental model
# You should now think of an object as:
# OBJECT
# │
# ├── Attributes → what it HAS
# │
# └── Methods    → what it DOES

# For an AI model:
# LLMModel
# │
# ├── name
# ├── temperature
# ├── max_tokens
# │
# ├── generate()
# ├── configure()
# └── describe()

# This is the foundation of OOP.
# One important distinction
# A function:
# def generate(prompt):    ...


# is standalone behavior.
# A method:
# class LLMModel:    def generate(self, prompt):        ...


# is behavior associated with an instance of LLMModel.
```

```python
class AIModel:
    def __init__(self, name, provider, temperature):
        self.name = name
        self.provider = provider
        self.temperature = temperature
    
    def describe(self):
        return f"{self.name} provided by {self.provider} runs at temperature {self.temperature}"
    

model1 = AIModel("Claude", "Anthropic", 0.7)
model1.describe()

model2 = AIModel("GPT", "OpenAI", 1.0)
model2.describe()
```

```python
class BankAccount:
    def __init__(self, balance):
        self.balance = balance

    def deposit(self, amount):
        self.balance += amount

    def withdraw(self, amount):
        if amount <= self.balance:
            self.balance -= amount
            return True
        
        return False
    
account = BankAccount(1000)

account.deposit(500)
print(account.balance)  # 1500

account.withdraw(300)
print(account.balance)  # 1200

account.withdraw(2000)
print(account.balance)  # 1200


# We're modifying an attribute that persists inside the object.
# Every subsequent method call sees the updated value.
# This is called state management.
```

```python
class ContextManager:
    def __init__(self, max_tokens):
        self.max_tokens = max_tokens
        self.used_tokens = 0

    def add_message(self, tokens):
        if tokens <= (self.max_tokens - self.used_tokens):
            self.used_tokens += tokens
            return True
        
        return False
    
    def remaining_tokens(self):
        return self.max_tokens - self.used_tokens
    
    def reset(self):
        self.used_tokens = 0

manager = ContextManager(8000)

print(manager.add_message(2500))
print(manager.add_message(3000))
print(manager.remaining_tokens())

print(manager.add_message(4000))
print(manager.remaining_tokens())

manager.reset()
print(manager.remaining_tokens())

print(manager.add_message(8000))
print(manager.remaining_tokens())
```

Class Attributes vs Instance Attributes

```python
# Python also supports attributes shared across all instances of a class.

class ToolTracker:
    total_calls = 0 # class attributes
    def __init__(self, name):
        self.name = name # instance attributes
        self.calls = 0

    def execute(self):
        self.calls += 1
        ToolTracker.total_calls += 1

search = ToolTracker("Web Search")
calculator = ToolTracker("Calculator")

search.execute()
search.execute()
calculator.execute()

print(search.calls)
print(calculator.calls)
print(ToolTracker.total_calls)

calculator.execute()

print(search.calls)
print(calculator.calls)
print(ToolTracker.total_calls)
```

Instance Methods vs Class Methods vs Static Methods

```python
# 1. Instance methods - self
# These operate on a particular object.

class Document:
    def __init__(self, content):
        self.content = content
    
    def word_count(self):
        return len(self.content.split())

doc = Document("Python makes AI development easier")
print(doc.word_count())
```

```python
# 2. Class methods - cls
# Sometimes a method needs to work with the class itself, rather than an individual object.

class Document:
    def __init__(self, content):
        self.content = content

    @classmethod
    def from_lines(cls, lines):
        combined_text = "\n".join(lines)
        return cls(combined_text)
    
doc = Document.from_lines([
    "Introduction to AI",
    "Machine Learning",
    "Deep Learning"
])

print(doc.content)


# What happened?
# Normally, we create an object like:

# doc = Document("Some text")


# But here:

# doc = Document.from_lines([...])


# We used a class method to create an object from a different input format.
# This is called an alternative constructor.
# Notice:

# return cls(combined_text)


# cls refers to the class (Document in this example).
# Class methods can also read and modify class attributes.
```
